## Part 2 — Baseline Agent-Based Model (Synthetic Grid)

Core ABM on a 20×20 grid (~1 km² urban block). Agents arrive via a Poisson process; app-users receive a direct match within a radius of 4 grid units. Key output metrics: mean search time (minutes), distance travelled (km), emissions (kg CO₂), and give-up rate (%).

**Parameters are documented inline. See v2 for sensitivity analysis on these values.**

```python
# ============================================================
#  ParkShare Madrid — Agent-Based Simulation
#  Ready to run in Google Colab (copy-paste this entire file)
#  Output: 4 charts (PNG) + results table (CSV) + parameter table (PNG)
# ============================================================
#
#  HOW TO RUN:
#  1. Go to colab.research.google.com
#  2. New notebook → paste this code into a cell
#  3. Run the cell (Shift+Enter)
#  4. Download files from the Files panel (left sidebar)
#
# ============================================================

# Install dependencies (already available in Colab, just in case)
import subprocess
subprocess.run(["pip", "install", "matplotlib", "pandas", "numpy", "-q"])

import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.table import Table
import pandas as pd
import random

# ── Reproducibility ───────────────────────────────────────────────────────────
SEED = 42

# ============================================================
#  SIMULATION PARAMETERS
#  Each parameter is justified below — see also parameter table
# ============================================================

GRID        = 20      # 20x20 grid ≈ 1 km² urban block (Monforte de Lemos area)
N_SPOTS     = 55      # number of parking spots on the grid
INIT_OCC    = 0.94    # 94% initial occupancy — peak hour SER zone (Madrid)
N_RUNS      = 30      # Monte Carlo runs per scenario (statistical robustness)
N_STEPS     = 600     # simulation steps per run
ARR_RATE    = 0.10    # Poisson arrival rate — new drivers per step
DEP_PROB    = 0.006   # probability a parked car departs each step (slow turnover)
RADIUS      = 4       # ParkShare match radius ≈ 200–300m in grid scale
S2M         = 0.60    # step-to-minute conversion (calibrated to baseline ≈ 8 min)
S2KM        = 0.030   # step-to-km conversion
COST_PRIV   = 3.0     # €/hr private parking (Ayuntamiento de Madrid tariff)
COST_P2P    = 3.0     # flat €3 ParkShare fee (ParkShare business model)
CO2_KM      = 0.12    # kg CO₂ per km (European Environment Agency, average fleet)
MAX_MIN     = 40.0    # give-up threshold in minutes (Monforte de Lemos baseline)

RATES  = [0.0, 0.10, 0.30, 0.60, 1.0]
LABELS = ['0%', '10%', '30%', '60%', '100%']

# ParkShare colour palette
PS_GRAY  = '#c8d8e4'
PS_LIGHT = '#b0cfe8'
PS_MID   = '#5ba3d0'
PS_DARK  = '#2a6496'
BAR_C    = [PS_GRAY, PS_LIGHT, PS_LIGHT, PS_MID, PS_DARK]

mpl.rcParams.update({
    'font.family':'sans-serif','font.size':10,
    'axes.titlesize':11,'axes.titleweight':'bold','axes.titlepad':12,
    'axes.labelsize':9,'axes.labelcolor':'#666',
    'xtick.labelsize':9,'ytick.labelsize':9,
    'xtick.color':'#888','ytick.color':'#888',
    'axes.spines.top':False,'axes.spines.right':False,
    'axes.spines.left':False,'axes.spines.bottom':False,
    'axes.grid':True,'axes.grid.axis':'y',
    'grid.color':'#e8e8e8','grid.linewidth':0.6,
    'xtick.bottom':False,'ytick.left':False,
    'figure.dpi':300,'figure.facecolor':'white',
})

SRC  = 'ParkShare ABM simulation  ·  30 Monte Carlo runs  ·  Calibrated to Madrid peak-hour conditions'
KW_S = dict(fontsize=7.5, color='#999', ha='left')
KW_L = dict(ha='center', va='bottom', fontsize=9, fontweight='bold', color='#333')

# ============================================================
#  CORE SIMULATION FUNCTION
# ============================================================

def run_episode(rate, seed):
    rng    = random.Random(seed)
    np_rng = np.random.default_rng(seed)

    all_pos  = [(x, y) for x in range(GRID) for y in range(GRID)]
    spot_pos = rng.sample(all_pos, N_SPOTS)
    spots    = [{'x':x, 'y':y, 'occ': rng.random() < INIT_OCC, 'res': None}
                for x, y in spot_pos]

    drivers, completed, nid = [], [], 0

    for _ in range(N_STEPS):

        # 1. New arrivals (Poisson process)
        for _ in range(np_rng.poisson(ARR_RATE)):
            drivers.append({'id':nid, 'app': rng.random() < rate,
                'x': rng.uniform(0, GRID), 'y': rng.uniform(0, GRID),
                'state':'searching', 'ss':0, 'ds':0, 'target':None})
            nid += 1

        # 2. Random departures — free up spots
        for s in spots:
            if s['occ'] and s['res'] is None and rng.random() < DEP_PROB:
                s['occ'] = False

        # 3. Update each driver
        done = []
        for d in drivers:
            if d['state'] in ('parked', 'gave_up'):
                done.append(d)
                continue

            d['ss'] += 1
            sm = d['ss'] * S2M

            # Give up after MAX_MIN minutes
            if sm > MAX_MIN:
                d['state'] = 'gave_up'
                if d['target']:
                    d['target']['res'] = None
                    d['target'] = None
                done.append(d)
                continue

            # APP USERS: find nearest free spot within RADIUS
            if d['app'] and d['state'] == 'searching':
                cands = [s for s in spots
                         if not s['occ'] and s['res'] is None
                         and abs(s['x']-d['x']) <= RADIUS
                         and abs(s['y']-d['y']) <= RADIUS]
                if cands:
                    best = min(cands, key=lambda s: (s['x']-d['x'])**2+(s['y']-d['y'])**2)
                    best['res'] = d['id']
                    d['target'] = best
                    d['state']  = 'matched'

            # MATCHED: navigate directly to reserved spot
            if d['state'] == 'matched' and d['target']:
                dx   = d['target']['x'] - d['x']
                dy   = d['target']['y'] - d['y']
                dist = (dx**2 + dy**2) ** 0.5
                if dist < 0.55:
                    d['target']['occ'] = True
                    d['target']['res'] = None
                    d['x'] = d['target']['x']
                    d['y'] = d['target']['y']
                    d['state']  = 'parked'
                    d['target'] = None
                    done.append(d)
                else:
                    spd = min(1.1, dist)
                    d['x'] += dx/dist * spd
                    d['y'] += dy/dist * spd
                    d['ds'] += 1
                continue

            # NO-APP USERS: random walk, park if adjacent spot free
            if d['state'] == 'searching' and not d['app']:
                adj = [s for s in spots
                       if not s['occ'] and s['res'] is None
                       and abs(s['x']-d['x']) < 1.0
                       and abs(s['y']-d['y']) < 1.0]
                if adj:
                    s = adj[0]
                    s['occ'] = True
                    d['x'] = s['x']
                    d['y'] = s['y']
                    d['state'] = 'parked'
                    done.append(d)
                    continue
                # Random walk
                d['x'] += rng.uniform(-1.1, 1.1)
                d['y'] += rng.uniform(-1.1, 1.1)
                d['x']  = max(0, min(GRID-1, d['x']))
                d['y']  = max(0, min(GRID-1, d['y']))
                d['ds'] += 1

        # 4. Record completed drivers
        for d in done:
            if d['ss'] > 0:
                sm  = d['ss'] * S2M
                dkm = d['ds'] * S2KM
                gu  = d['state'] == 'gave_up'
                cost = (COST_PRIV * (sm/60 + 0.5)) if gu else (COST_P2P if d['app'] else 0.5)
                completed.append({'app':d['app'], 'sm':sm, 'dkm':dkm,
                                  'cost':cost, 'co2':dkm*CO2_KM, 'gu':gu})
        drivers = [d for d in drivers if d['state'] not in ('parked','gave_up')]

    return completed

# ============================================================
#  RUN ALL SCENARIOS (Monte Carlo)
# ============================================================

print("Running ParkShare ABM simulation...")
print("Scenarios: 0%, 10%, 30%, 60%, 100% adoption\n")

results = {}
for rate in RATES:
    pool = []
    for run in range(N_RUNS):
        pool.extend(run_episode(rate, SEED + run * 13))
    results[rate] = pool
    t = np.mean([r['sm']  for r in pool])
    g = 100 * np.mean([r['gu'] for r in pool])
    print(f"  Adoption {int(rate*100):3d}%  |  n={len(pool):4d} drivers  |"
          f"  avg search = {t:.1f} min  |  give-up = {g:.1f}%")

# ============================================================
#  BUILD SUMMARY TABLE
# ============================================================

summary = []
for rate, label in zip(RATES, LABELS):
    d = results[rate]
    summary.append({
        'Adoption rate':          label,
        'Avg search time (min)':  round(np.mean([r['sm']   for r in d]), 1),
        'Avg distance (km)':      round(np.mean([r['dkm']  for r in d]), 3),
        'Avg cost (€)':           round(np.mean([r['cost'] for r in d]), 2),
        'Avg CO₂ (kg)':           round(np.mean([r['co2'] for r in d]),  4),
        'Give-up rate (%)':       round(100*np.mean([r['gu'] for r in d]), 1),
    })

df = pd.DataFrame(summary)
print("\n── Results table ───────────────────────────────────────")
print(df.to_string(index=False))
df.to_csv('parkshare_results.csv', index=False)
print("\nSaved: parkshare_results.csv")

# ============================================================
#  HELPER: save a bar chart
# ============================================================

def save_chart(fname, vals, ylabel, title, source, fmt, pad=0.025, extra=''):
    fig, ax = plt.subplots(figsize=(6.5, 3.6))
    bars = ax.bar(LABELS, vals, color=BAR_C, width=0.5, zorder=3, edgecolor='white')
    for b, v in zip(bars, vals):
        ax.text(b.get_x()+b.get_width()/2, b.get_height()+max(vals)*pad,
                fmt.format(v), **KW_L)
    ax.set_ylabel(ylabel)
    ax.set_xlabel('ParkShare adoption rate', labelpad=8)
    ax.set_ylim(0, max(vals) * 1.20)
    ax.set_title(title, loc='center')
    fig.text(0.02, -0.02, source + extra, **KW_S)
    fig.tight_layout()
    fig.savefig(fname, bbox_inches='tight', dpi=300)
    plt.close()
    print(f"Saved: {fname}")

# ============================================================
#  CHARTS A–D
# ============================================================

times = [s['Avg search time (min)'] for s in summary]
dists = [s['Avg distance (km)']     for s in summary]
costs = [s['Avg cost (€)']          for s in summary]
co2s  = [s['Avg CO₂ (kg)']          for s in summary]

print("\n── Saving charts ───────────────────────────────────────")

save_chart('sim_A_search_time.png', times, 'Minutes',
           'Average parking search time by adoption rate',
           SRC, '{:.1f} min')

save_chart('sim_B_distance.png', dists, 'Kilometres',
           'Average distance traveled during parking search',
           SRC, '{:.3f} km')

save_chart('sim_C_cost.png', costs, 'Euros (€)',
           'Average parking cost per driver by adoption rate',
           SRC, '€{:.2f}',
           extra='  ·  Private parking = €3/hr  ·  P2P transaction = €3 flat')

save_chart('sim_D_co2.png', co2s, 'kg CO₂',
           'Average CO₂ emissions during parking search',
           SRC, '{:.4f} kg',
           extra='  ·  CO₂ proxy: 0.12 kg/km (EEA average fleet)')

# ============================================================
#  PARAMETER JUSTIFICATION TABLE
# ============================================================

param_data = [
    ['GRID',      '20 × 20',  'Grid cells',  'Represents ~1 km² urban block (Monforte de Lemos)'],
    ['N_SPOTS',   '55',       'Spots',       'Approx. 1 spot per 7 grid cells — dense urban street'],
    ['INIT_OCC',  '0.94',     '—',           'Peak-hour SER zone occupancy, Madrid (Ayuntamiento de Madrid)'],
    ['ARR_RATE',  '0.10',     'Drivers/step','Poisson process; calibrated to congested peak conditions'],
    ['DEP_PROB',  '0.006',    'Prob./step',  'Slow turnover — consistent with peak-hour low churn'],
    ['RADIUS',    '4',        'Grid cells',  'ParkShare 300m match radius (ParkShare UX design spec)'],
    ['S2M',       '0.60',     'Min/step',    'Calibrated so baseline avg ≈ 8.1 min (Shoup, 2006)'],
    ['S2KM',      '0.030',    'Km/step',     'Calibrated to grid cell ≈ 50m physical distance'],
    ['MAX_MIN',   '40',       'Minutes',     'Avg. cruising time before giving up (ParkShare paper — Monforte)'],
    ['COST_PRIV', '3.0',      '€/hr',        'Private parking tariff, central Madrid (Ayuntamiento de Madrid)'],
    ['COST_P2P',  '3.0',      '€/transaction','ParkShare flat fee (ParkShare business model spec)'],
    ['CO2_KM',    '0.12',     'kg CO₂/km',   'Average passenger car emissions (EEA, 2023)'],
    ['N_RUNS',    '30',       'Runs',        'Monte Carlo robustness — standard for ABM studies'],
]

col_labels = ['Parameter', 'Value', 'Unit', 'Justification']
col_widths  = [0.13, 0.08, 0.13, 0.66]

fig, ax = plt.subplots(figsize=(11, 5.2))
ax.axis('off')

tbl = ax.table(
    cellText=param_data,
    colLabels=col_labels,
    cellLoc='left',
    loc='center',
    bbox=[0, 0, 1, 1]
)

tbl.auto_set_font_size(False)
tbl.set_fontsize(8.5)

# Style header
for col in range(4):
    cell = tbl[0, col]
    cell.set_facecolor('#2a6496')
    cell.set_text_props(color='white', fontweight='bold')
    cell.set_height(0.10)

# Style rows
for row in range(1, len(param_data)+1):
    bg = '#f0f7fd' if row % 2 == 0 else 'white'
    for col in range(4):
        cell = tbl[row, col]
        cell.set_facecolor(bg)
        cell.set_edgecolor('#dde8f0')
        cell.set_height(0.08)

# Set column widths
for col, w in enumerate(col_widths):
    for row in range(len(param_data)+1):
        tbl[row, col].set_width(w)

ax.set_title('Simulation parameter justification table',
             fontsize=11, fontweight='bold', pad=14, loc='left')
fig.text(0.02, 0.01,
         'EEA = European Environment Agency  ·  Ayuntamiento = Madrid City Council open data  ·  '
         'Calibrated = adjusted so simulated baseline matches empirical benchmark (Shoup 2006 / Hemmatpour et al. 2025)',
         fontsize=7, color='#999')

fig.tight_layout()
fig.savefig('sim_E_parameter_table.png', bbox_inches='tight', dpi=300)
plt.close()
print("Saved: sim_E_parameter_table.png")

print("""
════════════════════════════════════════════════
  ALL FILES SAVED — download from Files panel:

  sim_A_search_time.png   → Chart: search time
  sim_B_distance.png      → Chart: distance
  sim_C_cost.png          → Chart: cost
  sim_D_co2.png           → Chart: CO₂
  sim_E_parameter_table.png → Parameter table
  parkshare_results.csv   → Full results table
════════════════════════════════════════════════
""")
```
